 # Medical Insurance Cost Dataset Overview

## Introduction

This dataset contains information about individual health insurance beneficiaries and the corresponding medical insurance 
charges incurred by each person. 
The objective is to understand the factors that influence healthcare costs and to develop predictive models that 
estimate insurance expenses.



## Dataset Structure

The dataset contains **1,338 records** and **7 variables**, including demographic, lifestyle, and cost-related information.

| Variable | Type | Description |
|-----------|------|-------------|
| age | Numeric | Age of the primary beneficiary |
| sex | Categorical | Gender of the insurance beneficiary |
| bmi | Numeric | Body Mass Index, an indicator of body fat based on height and weight |
| children | Numeric | Number of dependents covered by the insurance plan |
| smoker | Categorical | Indicates whether the individual is a smoker |
| region | Categorical | Residential region in the United States |
| charges | Numeric | Annual medical insurance charges billed by the insurer |

---

## Business Objective

The primary goal of this analysis is to identify the factors that drive medical insurance costs and assess
how demographic and lifestyle characteristics affect insurance expenses.

Key questions include:

- Does age influence insurance charges?
- How does smoking status impact healthcare costs?
- Is BMI associated with higher medical expenses?
- Do costs vary across different regions?
- Can insurance charges be accurately predicted using available customer information?

---

## Initial Observations

Several variables are expected to have a significant impact on insurance costs:

- **Age:** Older individuals may require more healthcare services, resulting in higher charges.
- **BMI:** Higher BMI values are often associated with increased health risks and medical expenditures.
- **Smoking Status:** Smokers are generally expected to incur substantially higher healthcare costs.
- **Children:** The number of dependents may influence healthcare utilization patterns.
- **Region:** Geographic differences may contribute to variations in medical costs.

---

## Analytical Approach

The project will follow these steps:

1. Data quality assessment and cleaning
2. Exploratory Data Analysis (EDA)
3. Identification of key cost drivers
4. Statistical analysis of relationships between variables
5. Development of predictive models for insurance charges
6. Business interpretation and pricing recommendations

---

## Expected Outcome

The analysis aims to provide actionable insights into the determinants of medical insurance costs and support data-driven 
decision-making for insurance pricing, risk assessment, and customer segmentation.

Understanding these relationships can help insurers better estimate future costs, identify high-risk profiles, 
and design more accurate pricing strategies.

```python
# Import library
import pandas  as pd #Data manipulation
import numpy as np #Data manipulation
import matplotlib.pyplot as plt # Visualization
import seaborn as sns #Visualization
plt.rcParams['figure.figsize'] = [8,5]
plt.rcParams['font.size'] =14
plt.rcParams['font.weight']= 'bold'
plt.style.use('seaborn-whitegrid')
```

# ------------------------------------------
# 1. Load Data
# ------------------------------------------

```python
# Import dataset
#path ='dataset/'

df = pd.read_csv('insurance.csv')
print('\nNumber of rows and columns in the data set: ',df.shape)
print('')

#Lets look into top few rows and columns in the dataset
df.head()
```

# ------------------------------------------
# 2. Data Structure Review
# ------------------------------------------

```python
print("\nDataset Information:")
print(df.info())

print("\nColumn Names:")
print(df.columns.tolist())
```

# ------------------------------------------
# 3. Check Missing Values
# ------------------------------------------

```python
print("\nMissing Values:")
missing_values = df.isnull().sum()

missing_summary = pd.DataFrame({
    'Missing_Count': missing_values,
    'Missing_Percentage': round((missing_values / len(df)) * 100, 2)
})

print(missing_summary)
```

# ------------------------------------------
# 4. Check Duplicate Records
# ------------------------------------------


```python
duplicate_count = df.duplicated().sum()

print(f"\nNumber of Duplicate Records: {duplicate_count}")

# Remove duplicates if any
df = df.drop_duplicates()

print(f"Dataset Shape After Removing Duplicates: {df.shape}")
```

# ------------------------------------------
# 5. Review Data Types
# ------------------------------------------

```python
print("\nData Types:")
print(df.dtypes)
```

# ------------------------------------------
# 6. Descriptive Statistics
# ------------------------------------------

```python
print("\nNumerical Summary:")
print(df.describe())

print("\nCategorical Summary:")
print(df.describe(include='object'))
```

# ------------------------------------------
# 7. Check Unique Values in Categorical Variables
# ------------------------------------------

```python
categorical_cols = ['sex', 'smoker', 'region']

for col in categorical_cols:
    print(f"\nUnique values in '{col}':")
    print(df[col].value_counts())
```

# ------------------------------------------
# 8. Check for Invalid Values
# ------------------------------------------

```python
print("\nChecking for Invalid Values...")

# Age should be positive
invalid_age = df[df['age'] <= 0]

# BMI should be positive
invalid_bmi = df[df['bmi'] <= 0]

# Children should not be negative
invalid_children = df[df['children'] < 0]

# Charges should be positive
invalid_charges = df[df['charges'] <= 0]

print(f"Invalid Age Records: {len(invalid_age)}")
print(f"Invalid BMI Records: {len(invalid_bmi)}")
print(f"Invalid Children Records: {len(invalid_children)}")
print(f"Invalid Charges Records: {len(invalid_charges)}")
```

# ------------------------------------------
# 9. Outlier Detection (IQR Method)
# ------------------------------------------

```python
numeric_cols = ['age', 'bmi', 'children', 'charges']

print("\nPotential Outliers:")

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} outliers")
```

# ------------------------------------------
# 10. Standardize Text Values
# ------------------------------------------

```python
for col in categorical_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
        .str.lower()
    )

# Verify results
print("\nCleaned Categorical Values:")

for col in categorical_cols:
    print(f"\n{col}")
    print(df[col].unique())
```

# ------------------------------------------
# 11. Final Data Quality Report
# ------------------------------------------

```python
print("\n========== DATA QUALITY REPORT ==========")
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
print(f"Missing Values: {df.isnull().sum().sum()}")
print(f"Duplicate Records: {df.duplicated().sum()}")
```
